```python
import pandas as pd
import numpy as np

```

```python
df = pd.read_csv('/content/Telco-Customer-Churn.csv')
```

```python
df.isnull().sum()
```

```python
df.drop('customerID', axis =1, inplace = True)
```

```python
df.describe()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df.describe()
```

```python
df.columns
```

```python
df.shape
```

```python
df = df.dropna(subset=['TotalCharges'])
df.reset_index(drop=True, inplace=True)
```

```python
df.shape
```

```python
# DataPreprocessing

1. using Label encoder to convert categorical values to numerical values
```

```python
from sklearn.preprocessing import LabelEncoder

cols = ['gender', 'Partner', 'Dependents','PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity','OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV','StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod','Churn']
d = {}

for i in cols:
  le = LabelEncoder()
  df[i] = le.fit_transform(df[i])

  classes = le.classes_
  print("classes",classes)

  map_d = dict(zip(classes, range(len(classes))))
  print("mapping",map_d)
  d.update(map_d)

```

```python
df.head(5)
```

```python
# Stadardisation: using Standard scaler
    # selecting feature, target column
```

```python
x = df.iloc[:,0:-1]
y = df['Churn']
```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()

ft = ss.fit_transform(x)
x_df = pd.DataFrame(ft)
x_df.head(3)
```

```python
# Modeling:

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x_df,y, test_size = 0.2,random_state = 30)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
# Classification problem
```

```python
# Model 1 : Logistic Regression classifier


from sklearn.linear_model import LogisticRegression

m1 = LogisticRegression( class_weight='balanced')
m1.fit(x_train,y_train)
```

```python
pred1 = m1.predict(x_test)
pred1
```

```python
#

from sklearn.metrics import accuracy_score, confusion_matrix,classification_report
accuracy_score(y_test,pred1)
```

```python
print(classification_report(y_test,pred1))
```

```python
# Model 2: Desicion Tree Classifier

from sklearn.tree import DecisionTreeClassifier

m2 = DecisionTreeClassifier(class_weight='balanced',max_depth=10, min_samples_split=5, random_state=42)
m2.fit(x_train,y_train)
```

```python
pred2 = m2.predict(x_test)
pred2
```

```python
accuracy_score(y_test,pred2)
```

```python
print(classification_report(y_test,pred2))
```

```python
# Model 3: RandomForest Classifier

from sklearn.ensemble import RandomForestClassifier

m3  = RandomForestClassifier(class_weight='balanced',max_depth=12, min_samples_split=6, random_state=50)
m3.fit(x_train,y_train)
```

```python
pred3 = m3.predict(x_test)
pred3
```

```python
accuracy_score(y_test,pred3)
```

```python
print(classification_report(y_test,pred3))
```

```python
# missing a customer who is likely to churn (a false
# negative) is more costly than mistakenly predicting churn for a customer who
# actually stays (a false positive).

# the best model is Logistic regression classifier with (recall = 0.78), meaning it's catching ~78% of churners
```
